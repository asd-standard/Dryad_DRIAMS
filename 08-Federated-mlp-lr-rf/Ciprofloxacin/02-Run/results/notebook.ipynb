```python
from keras.datasets import mnist
from keras import layers
import keras
from matplotlib import pyplot as plt
import time
```

```python
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

```python
model = keras.Sequential(
    [
        layers.Dense(512, activation='relu'),
        layers.Dense(10, activation='softmax'),
    ]
)
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'],
)
```

```python
#keras want flat 1D images with float32 values mnist images are 2D uint8
```

```python
train_images = train_images.reshape((60000, 28 * 28))
train_images = train_images.astype("float32") / 255
test_images = test_images.reshape((10000, 28 * 28))
test_images = test_images.astype("float32") / 255
```

```python
model.fit(train_images, train_labels, epochs=10, batch_size=128)
```

```python
test_digits = test_images[0:10]
predictions = model.predict(test_digits)
```

```python
predictions[0]
```

```python
view_image = test_images[0].reshape(28, 28)
```

```python
one_image = test_images[0]
one_image = one_image.reshape(1, -1)
pred = model.predict(one_image)
```

```python
plotted_image = plt.imshow(view_image, cmap=plt.cm.binary)
plt.show()
```

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
```

```python
print(f"test_acc: {test_acc}")
```
